```python
import pandas as pd
```

```python
df = pd.read_csv("uber.csv")
print(f"Original rows: {len(df):,}")
```

```python
df = df.drop(columns=["key"], errors="ignore")
df = df.reset_index(drop=True)
```

```python
df = df.dropna()
```

```python
df = df[(df["fare_amount"] >= 2.5) & (df["fare_amount"] <= 300)]
df = df[(df["passenger_count"] >= 1) & (df["passenger_count"] <= 6)]
df = df[
    (df["pickup_latitude"].between(40.0, 42.0)) &
    (df["pickup_longitude"].between(-75.0, -72.0)) &
    (df["dropoff_latitude"].between(40.0, 42.0)) &
    (df["dropoff_longitude"].between(-75.0, -72.0))
]
```

```python
df["pickup_datetime"] = pd.to_datetime(df["pickup_datetime"])
df["year"] = df["pickup_datetime"].dt.year
df["month"] = df["pickup_datetime"].dt.month
df["hour"] = df["pickup_datetime"].dt.hour
df["day_name"] = df["pickup_datetime"].dt.day_name()
```

```python
df.to_csv("uber_cleaned.csv", index=False)
```

```python
print(f"Cleaned rows: {len(df):,}")
print(f"Rows removed: {200000 - len(df):,}")
print("\nFirst 5 rows of cleaned data:")
print(df.head())
```
